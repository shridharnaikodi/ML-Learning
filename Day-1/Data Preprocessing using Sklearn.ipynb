```python
# Whenever you are working on sklearn classes, ensure your data is represented in the form of
# numpy array
```

```python
#Step1: Seperate data as features and label

features = data.iloc[:,[0,1,2]].values
label = data.iloc[:,[3]].values
```

```python
features
```

```python
label
```

```python
# Handling Missing Value ----- Imputation
features[:,[0]]
```

```python
#Step1: Import relevant library

from sklearn.impute import SimpleImputer

#Step2: Instantiate the SimpleImputer Object

siForCountry = SimpleImputer(missing_values=np.nan, strategy='most_frequent')

#Step3: Fit the data to the object (Internally it will calc the mode of the country column)

siForCountry.fit(features[:,[0]])

#Step4: Transform the data ---> In this case, transform means replace nan with the mode value calc above
features[:,[0]] = siForCountry.transform(features[:,[0]])
```

```python
features
```

```python
# Age --- median
#Step1: Import relevant library

from sklearn.impute import SimpleImputer

#Step2: Instantiate the SimpleImputer Object

siForAge = SimpleImputer(missing_values=np.nan, strategy='median')

#Step3: Fit the data to the object (Internally it will calc the mode of the country column)

siForAge.fit(features[:,[1]])

#Step4: Transform the data ---> In this case, transform means replace nan with the mode value calc above
features[:,[1]] = siForAge.transform(features[:,[1]])
```

```python
features
```

```python
# Salary ---- mean
#Step1: Import relevant library

from sklearn.impute import SimpleImputer

#Step2: Instantiate the SimpleImputer Object

siForSalary = SimpleImputer(missing_values=np.nan, strategy='mean')

#Step3: Fit the data to the object (Internally it will calc the mode of the country column)

siForSalary.fit(features[:,[2]])

#Step4: Transform the data ---> In this case, transform means replace nan with the mode value calc above
features[:,[2]] = siForSalary.transform(features[:,[2]])
```

```python
features
```

```python
# Sklearn also performs OHE using  OneHotEncoder class
```

```python
from sklearn.preprocessing import OneHotEncoder

oheForCountry = OneHotEncoder(sparse_output=False)

oheForCountry.fit(features[:,[0]]) #This will do the same OHE Process (Extract unique values and sort in asc order)


countryDummyVariable = oheForCountry.transform(features[:,[0]]) #Create Dummy Variables
```

```python
countryDummyVariable
```

```python
oheForCountry.categories_
```

```python
# Transform categorical to dummy

oheForCountry.transform(np.array(["Germany"]).reshape(-1,1))
```

```python
# Transform dummy to categorical

oheForCountry.inverse_transform(np.array([[0., 0., 1.]]))
```
